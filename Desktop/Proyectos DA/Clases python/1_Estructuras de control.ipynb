### Estructuras de Control en Programación

En programación, las estructuras de control son los bloques de construcción fundamentales que determinan el flujo de ejecución de un programa. Piensa en ellas como las señales de tráfico y las direcciones que le dices a tu programa para que siga. Sin ellas, un programa simplemente ejecutaría todas sus instrucciones de principio a fin, una tras otra, de forma lineal.

## Las estructuras de control permiten a los programas:

Tomar decisiones: Ejecutar un bloque de código u otro basándose en una condición (estructuras condicionales).

Repetir acciones: Ejecutar un bloque de código múltiples veces (estructuras de bucle o iterativas).

Manejar errores: Responder de forma controlada a situaciones inesperadas (estructuras de manejo de excepciones, que se verán más adelante).

### Tipos Principales de Estructuras de Control:

* Condicionales: 

Permiten que el programa evalúe una condición y ejecute diferentes bloques de código dependiendo de si esa condición es True (verdadera) o False (falsa). Son la base para la toma de decisiones en tu código.

* Iterativas (Bucles): 

Permiten que un bloque de código se repita un número determinado de veces o mientras una condición sea verdadera. Son esenciales para procesar colecciones de datos o realizar tareas repetitivas. (Se verán en detalle más adelante).

### 🧠 El Condicional if

El condicional if (del inglés "si") es la estructura de control condicional más básica y fundamental en Python.

Su propósito es ejecutar un bloque de código solo si una condición específica se evalúa como True.

### Sintaxis y Funcionamiento:

La sintaxis básica de la sentencia if es la siguiente:

``` Python
if condicion_a_evaluar:
    # Bloque de código a ejecutar si la condicion_a_evaluar es True
    # Todas estas líneas deben estar indentadas al mismo nivel
```     


### Puntos Clave del Funcionamiento de if:

- condicion_a_evaluar: 

Esta es una expresión que Python evaluará. El resultado de esta evaluación debe ser un valor booleano (True o False).

Las condiciones suelen formarse utilizando operadores de comparación (como >, <, ==, !=, >=, <=) y operadores lógicos (como and, or, not).

- Dos Puntos ( : ): 

Después de la condicion_a_evaluar, siempre debe ir un colon (:). Esto le indica a Python que a continuación viene un bloque de código.

- Indentation (Sangría): 

Este es uno de los aspectos más importantes y distintivos de Python. El bloque de código que pertenece a la sentencia if debe estar indentado. La indentación estándar en Python es de cuatro espacios. Todas las líneas dentro del mismo bloque if deben tener el mismo nivel de indentación. Si una línea no está indentada correctamente, Python generará un error IndentationError o la considerará fuera del bloque if.

- Flujo de Ejecución:

Si la condicion_a_evaluar se evalúa como True, el bloque de código indentado debajo del if se ejecuta.

Si la condicion_a_evaluar se evalúa como False, el bloque de código indentado se salta por completo, y el programa continúa la ejecución desde la primera línea de código que está al mismo nivel de indentación que la sentencia if.

El if es la puerta de entrada a la lógica de decisión en tus programas, permitiéndoles responder de manera dinámica a diferentes entradas y situaciones.

### 💻 EJEMPLOS PRÁCTICOS

````python
# --- Ejemplo 1: Condición Simple con un número ---
print("--- Ejemplo 1: Condición Simple con un número ---")
numero = 2

if numero > 5:
    print(f"El número {numero} es mayor que 5.")
    print("Este mensaje también se ejecuta porque la condición es verdadera.")

print("Este mensaje se ejecuta siempre, ya que está fuera del bloque if.")
````

````python
# --- Ejemplo 2: Condición Falsa ---
print("\n--- Ejemplo 2: Condición Falsa ---")
temperatura = int(input('introduce la temperatura'))

if temperatura < 20:
    print(f"La temperatura {temperatura}°C indica que hace frío.")
    
````

````python
# --- Ejemplo 3: Condición con texto (string) ---
print("\n--- Ejemplo 3: Condición con texto (string) ---")
nombre = str(input('introduce la temperatura'))

if nombre == "Alice": # '==' es el operador de comparación de igualdad
    print(f"¡Hola, {nombre}!")

otro_nombre = "Bob"
if otro_nombre == "Alice":
    print(f"¡Hola, {otro_nombre}! (Esto no se imprimirá)")
````

### ¿Qué es el Manejo de Errores?

En programación, un error (o excepción) es un evento que ocurre durante la ejecución de un programa y que interrumpe su flujo normal. Por ejemplo, ¿qué pasa si tu programa intenta dividir un número por cero? ¡Eso es matemáticamente imposible y el programa se detendrá! O, ¿qué pasa si pides al usuario que ingrese un número, pero él escribe texto? Python intentará convertir el texto a número y ¡boom!, otro error.

Para evitar que tu programa se "rompa" o se detenga inesperadamente, usamos el manejo de errores. Es como tener un "plan B" para cuando algo sale mal.

- Los bloques try y except:

Python nos proporciona una estructura especial para "intentar" ejecutar un código y, si falla, "capturar" el error y hacer algo al respecto en lugar de detener el programa.

### try (Intentar):

Dentro de este bloque, colocamos el código que sospechamos que podría causar un error.
Python intentará ejecutar todas las líneas de código dentro del try.

### except (Exceptuar / Capturar):

Si ocurre un error dentro del bloque try, Python detiene la ejecución del try y salta directamente al bloque except.

Dentro del except, colocamos el código que queremos ejecutar si y solo si ocurre el error. Esto podría ser imprimir un mensaje al usuario, registrar el error, o intentar una solución alternativa.

Podemos especificar qué tipo de error queremos "atrapar" (por ejemplo, ValueError, ZeroDivisionError). Si no especificamos ninguno, except capturará cualquier tipo de error. (Es buena práctica ser específico, pero para empezar, un except general funciona).

Sintaxis Básica:

```Python
try:
    # Código que podría causar un error
    # Por ejemplo: operaciones con entrada del usuario, división por cero, acceso a archivos
    print("Intentando ejecutar el código...")
    resultado = 10 / 0 # Esto causará un error de división por cero
    print(f"El resultado es: {resultado}") # Esta línea no se ejecutaría
except:
    # Código que se ejecuta si ocurre CUALQUIER error en el bloque try
    print("¡Oh no! Ha ocurrido un error.")
    print("Por favor, asegúrate de no dividir por cero o de ingresar datos válidos.")

print("El programa continúa después del manejo del error.") # El programa no se detiene

```

### Beneficios:

- Robustez: Tu programa no se "cae" por errores inesperados.

- Experiencia de Usuario: Puedes dar mensajes útiles y amigables al usuario en lugar de mostrarles un mensaje de error técnico de Python.

- Control: Tienes el control sobre cómo reacciona tu programa ante diferentes tipos de problemas.

Recuerda que try y except son herramientas para cuando esperas que algo pueda salir mal y quieres manejarlo elegantemente, no para evitar programar defensivamente en primer lugar (es decir, sigue validando entradas cuando sea posible antes de que ocurra un error).

### 💻 EJEMPLO PRÁCTICO: Conversión de Entrada del Usuario

Imagina que pides al usuario un número, pero ellos escriben "hola". 

Si intentas convertir "hola" a un número entero con int(), 

Python lanzará un ValueError. Con try-except podemos manejar esto.

````python
print("--- Ejemplo de try-except: Conversión de Entrada ---")

# Código que podría fallar
numero_str = input("Por favor, ingresa un número entero: ")

try:
    # Intentamos convertir la entrada a un entero
    numero_entero = int(numero_str)
    print(f"¡Excelente! Ingresaste el número: {numero_entero}")
    resultado = numero_entero * 2
    print(f"El doble de tu número es: {resultado}")

except ValueError:
    # Este bloque se ejecuta SOLAMENTE si ocurre un ValueError (ej. si el usuario no ingresa un número)
    print("¡Error! Eso no parece un número entero válido.")
    print("Por favor, intenta de nuevo ingresando solo dígitos (ej. 5, 10, -3).")
except:
    # Este es un except general para cualquier otro tipo de error que no sea ValueError
    print("Ocurrió un error inesperado. Por favor, contacta al soporte.")

print("Fin del programa. Gracias por participar.")
````

## Introducción a los Bucles en Python
¿Qué es un bucle?
Un bucle es una estructura de control que permite ejecutar una secuencia de instrucciones repetidamente hasta que se cumple una condición. En Python, los bucles se utilizan para automatizar tareas repetitivas.

Existen dos tipos principales de bucles en Python:

Bucle for y 
Bucle while

### 1- Bucle for en Python
El bucle for en Python se usa para iterar sobre una secuencia (que puede ser una lista, una tupla, un diccionario, un conjunto o una cadena de caracteres). Con cada iteración, el bucle toma el siguiente elemento de la secuencia y ejecuta el bloque de código asociado.

Sintaxis básica del bucle for:

```python
for variable in secuencia:

for i in secuencia:
````

variable: Es una variable que toma el valor de cada elemento de la secuencia en cada iteración.\
secuencia: Es la secuencia de elementos que queremos iterar.

### Ejemplo básico de un bucle for
Vamos a comenzar con un ejemplo simple donde imprimimos cada número de una lista:

`````python
numeros = [0, 2, 3, 4, 5] #lista

for i in numeros:
    #print(f'El número es:') #para mostrar el mismo mensaje 5 veces
    print(f'El número es:{i}' )#para mostrar que valor toma la variable
    
`````

### Explicación:
Creamos una lista llamada numeros que contiene los elementos [1, 2, 3, 4, 5].\
El bucle for toma cada número de la lista numeros y lo asigna a la variable numero.\
En cada iteración, imprimimos el valor de numero.

### 2- Bucle for con range()
El range() es una función útil en Python que se utiliza a menudo con el bucle for para generar una secuencia de números.

Ejemplo usando range():

`````python
for i in range(5):
    print('El valor de i es:',i)
`````

### Explicación:
range(5) genera una secuencia de números de 0 a 4.
En cada iteración, i toma un valor de la secuencia generada por range(5) y se imprime.

### 3- Bucle while en Python
El bucle while ejecuta un bloque de código mientras una condición sea verdadera. A diferencia del bucle for, el bucle while no depende de una secuencia, sino de una condición lógica.

Sintaxis básica del bucle while:
```Python
while condicion:
```    

condicion: Es una expresión booleana (True/False). El bucle sigue ejecutándose mientras esta condición sea True.

### Ejemplo básico de un bucle while
Vamos a crear un ejemplo en el que usamos un bucle while para contar de 0 a 5:

`````python
contador = 0

while contador <= 10:
    print('El contador es:', contador)
    contador += 1  # Incrementamos el contador en 1
`````

### Explicación:
Definimos una variable contador con un valor inicial de 0.
El bucle while sigue ejecutándose mientras contador sea menor o igual a 5.
Dentro del bucle, imprimimos el valor actual de contador y luego lo incrementamos en 1.

### Diferencia clave entre for y while
Bucle for: Se utiliza cuando se conoce de antemano el número de iteraciones. Itera sobre una secuencia o un rango.\
Bucle while: Se utiliza cuando no se conoce el número exacto de iteraciones y se basa en una condición que puede cambiar dentro del bucle.

### Ejemplo de while que podría no terminar (bucle infinito)
Es importante tener cuidado con los bucles while porque si la condición nunca se vuelve False, el bucle continuará indefinidamente, causando un "bucle infinito".

`````python
#ojo con las ejecuciones infinitas:
contador = 1

while contador <= 5:
    print('El contador es:', contador)
    # Aquí olvidamos incrementar el contador
`````

### Explicación:
En este ejemplo, el contador nunca se incrementa, por lo que contador <= 5 siempre es True y el bucle nunca termina.

### Conclusión:
for: Ideal para iterar sobre secuencias.

while: Ideal para repetir un bloque de código hasta que se cumpla una condición.

### Sentencias de Contro combinadas con Bucles:

- break : Detiene el bucle completamente.

- continue : Salta a la siguiente iteración del bucle.

- pass : No hace nada (se usa como marcador de posición).

### 🧾 Cuadro Explicativo de Bucles y Sentencias en Python

| Estructura     | ¿Qué hace?                                               | ¿Cuándo usarla?                                 |
|----------------|-----------------------------------------------------------|-------------------------------------------------|
| `for`          | Recorre una secuencia                                     | Con listas, rangos, strings, etc.               |
| `while`        | Repite mientras una condición sea verdadera               | Cuando no sabemos cuántas veces repetir         |
| `break`        | Sale del bucle completamente                              | Para detener el bucle por una condición         |
| `continue`     | Salta al siguiente ciclo del bucle                        | Para omitir ciertas iteraciones                 |
| `pass`         | No hace nada (placeholder o estructura vacía)             | Para dejar código incompleto temporalmente      |


### Ejemplos de Usos de sentencias:

`````python
empleados = [
    ('Juan', 'Gerente'),
    ('Ana', 'Analista'),
    ('Pedro', 'Desarrollador'),
    ('Lucía', 'Desarrollador'),
    ('Marta', 'Recursos Humanos'),
    ('Carlos', 'Gerente'),
    ('Sofía', 'Desarrollador'),
    ('Diego', 'Analista')
]
`````

### Uso del Metodo Brek cuando encontramos la poscion de Gerente

`````python
for nombre, posicion in empleados:
    if posicion == 'Recursos Humanos':
        print(f'{nombre} es un {posicion}, deteniendo búsqueda...')
        break  # Detiene el bucle cuando encontramos al primer "Gerente"
`````

### Uso del Metodo Continue cuando encontramos la poscion de Gerente

`````python
for nombre, posicion in empleados:
    if posicion == 'Gerente':
        continue  # Salta el resto del código en el bucle si la posición es "Desarrollador"
    print(f'{nombre} trabaja en {posicion}')
`````

### Uso del Metodo pass cuando encontramos la poscion de Gerente

`````python
for nombre, posicion in empleados:
    if posicion == 'Gerente':
        pass  # No hacemos nada aún con los "Gerente", pero el código no fallará
    else:
        print(f'{nombre} trabaja en {posicion}')
`````

### Condicionales if-else y elif

### El Condicional if-else:

Mientras que el condicional if nos permite ejecutar código solo si una condición es verdadera, 

el condicional if-else nos proporciona una bifurcación clara: 

ejecuta un bloque de código si la condición es verdadera, y un bloque de código alternativo si la condición es falsa. 

Esto es fundamental para definir comportamientos diferentes en dos escenarios mutuamente excluyentes.

### Sintaxis y Funcionamiento:

La sintaxis básica de la sentencia if-else es la siguiente:

```Python

if condicion_a_evaluar:
    # Bloque de código a ejecutar si la condicion_a_evaluar es True
else:
    # Bloque de código a ejecutar si la condicion_a_evaluar es False

```

### Puntos Clave del Funcionamiento de if-else:

- if y else son Excluyentes:

Solo uno de los dos bloques de código se ejecutará. Nunca se ejecutarán ambos, ni se omitirán ambos.

- else sin Condición: 

La palabra clave else no necesita una condición. Actúa como un "por defecto" o "en cualquier otro caso" si la condición if original es False.

- Indentation (Sangría): 

Al igual que con if, ambos bloques (if y else) deben estar correctamente indentados.
El else debe estar al mismo nivel de indentación que su if correspondiente.

### El Condicional elif (else if):

Cuando tienes más de dos posibles caminos o condiciones que evaluar, y estas condiciones son mutuamente excluyentes (es decir, solo una puede ser verdadera a la vez), el elif (abreviatura de "else if") es la herramienta perfecta. Te permite encadenar múltiples pruebas condicionales.

### Sintaxis y Funcionamiento:

La sintaxis básica de la sentencia if-elif-else es la siguiente:

```Python

if primera_condicion:
    # Código si la primera_condicion es True
elif segunda_condicion:
    # Código si la segunda_condicion es True (y la primera fue False)
elif tercera_condicion:
    # Código si la tercera_condicion es True (y las anteriores fueron False)
# ... puedes tener tantos elif como necesites ...
else:
    # Código si ninguna de las condiciones anteriores fue True
```


### Puntos Clave del Funcionamiento de elif:

- Evaluación en Cascada: 

Python evalúa las condiciones de arriba hacia abajo, en el orden en que aparecen:

Primero, se evalúa la condición del if. Si es True, se ejecuta su bloque de código y el resto de la estructura elif-else se salta por completo.

Si la condición del if es False, entonces se evalúa la primera condición elif. Si es True, se ejecuta su bloque de código y el resto de la estructura se salta.

Este proceso continúa con cada elif hasta que se encuentra una condición True o hasta que se llega al else.

- Solo se Ejecuta un Bloque:

Una vez que se encuentra una condición True y se ejecuta su bloque de código, el resto de la cadena elif y el bloque else (si existe) se ignoran. Esto significa que solo se ejecutará un único bloque de código dentro de toda la estructura if-elif-else.

- Else es Opcional: 

El bloque else al final de una cadena if-elif es opcional. Si lo omites y ninguna de las condiciones if o elif es verdadera, simplemente no se ejecutará ningún bloque de código dentro de esa estructura condicional.

- Orden de las Condiciones: 

El orden de tus condiciones elif es crucial. Si tienes condiciones que se superponen, asegúrate de que la condición más específica o la que quieres que se evalúe primero esté más arriba en la estructura.

### Las estructuras if-else y if-elif-else son la columna vertebral de la lógica de decisión en tus programas, permitiéndote manejar una amplia gama de escenarios y rutas de ejecución. ### 

### 💻 EJEMPLOS PRÁCTICOS

````python
# --- Ejemplo 1: if-else para Par o Impar ---
print("--- Ejemplo 1: if-else para Par o Impar ---")
numero = 8

if numero % 2 == 0: # El operador '%' es el módulo (resto de la división)
    print(f"El número {numero} es par.")
else:
    print(f"El número {numero} es impar.")

````

````python
# --- Ejemplo 2: if-else para Verificación de Mayoría de Edad ---
print("\n--- Ejemplo 2: if-else para Verificación de Mayoría de Edad ---")
edad = 20

if edad >= 18:
    print("Eres mayor de edad.")
else:
    print("Eres menor de edad.")
````

````python
# --- Ejemplo 3: if-elif-else para Calificaciones ---
print("\n--- Ejemplo 3: if-elif-else para Calificaciones ---")
puntuacion = int(input())

if puntuacion >= 90:  
    print("Calificación: A (Excelente)")
elif puntuacion >= 80: # Se ejecuta si puntuacion < 90 y puntuacion >= 80
    print("Calificación: B (Notable)")
elif puntuacion >= 70: # Se ejecuta si puntuacion < 80 y puntuacion >= 70
    print("Calificación: C (Bueno)")
elif puntuacion >= 60: # Se ejecuta si puntuacion < 70 y puntuacion >= 60
    print("Calificación: D (Aprobado)")
else: # Se ejecuta si puntuacion < 60
    print("Calificación: F (Reprobado)")
````

````python
# --- Ejemplo 4: if-elif-else con cadenas de texto y entrada del usuario ---
print("\n--- Ejemplo 4: if-elif-else con cadenas de texto y entrada del usuario ---")
dia_semana = input("¿Qué día de la semana es hoy? ").lower() # Convertimos a minúsculas para facilitar la comparación

if dia_semana == "lunes":
    print("¡Ánimo con la semana!")
elif dia_semana == "viernes":
    print("¡Casi es fin de semana!")
elif dia_semana == "sábado" or dia_semana == "domingo": # Usando operador lógico 'or'
    print("¡Disfruta el fin de semana!")
else:
    print("Es un día de semana normal.")

````

````python
# --- Ejemplo 5: Orden de las condiciones es importante ---
print("\n--- Ejemplo 5: Orden de las condiciones es importante ---")
valor = 30

# ¡CUIDADO! Esta estructura podría dar resultados inesperados si el orden es incorrecto
# # Si tuviéramos:
# if valor > 5:
#     print("Valor es mayor que 5")
# elif valor > 10: # Esta condición NUNCA se alcanzaría si valor > 5 ya es True y valor > 10
#     print("Valor es mayor que 10")

# La forma correcta es poner la condición más específica primero:
if valor > 10:
    print("El valor es mayor que 10")
elif valor > 5: # Solo se evalúa si 'valor > 10' fue False
    print("El valor es mayor que 5 (pero no mayor que 10)")
else:
    print("El valor es 5 o menor.")
````

### ¿Qué es la Indexación?
La indexación es un concepto fundamental en la programación que te permite acceder a elementos específicos dentro de una secuencia de datos. Imagina una fila de asientos en un cine: cada asiento tiene un número, y la indexación es como usar ese número para encontrar un asiento en particular. En Python, las secuencias de datos incluyen estructuras como listas, tuplas y cadenas de texto (strings).

### ¿Cómo funciona la indexación?
En Python, la indexación tiene algunas características clave:

Índices basados en cero: La característica más importante de la indexación en Python es que los índices siempre comienzan en 0. Esto significa que el primer elemento de una secuencia está en la posición 0, el segundo en la 1, y así sucesivamente. Si una secuencia tiene n elementos, el último elemento estará en la posición n-1.

### Ejemplo: 

En la lista ['manzana', 'banana', 'cereza'], 'manzana' está en el índice 0, 'banana' en el 1, y 'cereza' en el 2.

Índices positivos: 
Se utilizan para acceder a elementos desde el principio de la secuencia hacia el final. Son los más comunes y directos.

### Ejemplo: 

Si tienes una cadena de texto "Hola", cadena[0] te dará 'H', cadena[1] te dará 'o', y así sucesivamente.
Índices negativos: Python también te permite usar índices negativos, lo cual es muy útil para acceder a elementos desde el final de la secuencia sin necesidad de conocer su longitud.

-1 se refiere al último elemento.

-2 se refiere al penúltimo elemento.

Y así sucesivamente.

### Ejemplo:

En la lista ['rojo', 'verde', 'azul'], 
lista[-1] te dará 'azul', 
y lista[-2] te dará 'verde'.

### ¿Por qué es importante la indexación?
La indexación es crucial porque te permite:

### Acceder a elementos individuales: 

Recuperar un dato específico de una secuencia.

### Modificar elementos: 

En el caso de estructuras de datos mutables como las listas, puedes usar la indexación para cambiar el valor de un elemento.

### Realizar operaciones específicas: 

Muchos algoritmos y operaciones dependen de poder acceder a elementos por su posición.

### Iterar sobre secuencias: 

Aunque no se usa directamente en un bucle for-each, la indexación es la base para bucles que necesitan acceder a elementos por su posición.

### Limitaciones y errores comunes:

Es importante tener en cuenta que si intentas acceder a un índice que está fuera del rango de la secuencia (ya sea un índice positivo demasiado grande o un índice negativo demasiado pequeño), Python generará un error IndexError.

### 📚✨ Vamos a ver algunos casos ✨📚

```python
# 1. Acceso por índice positivo
mi_lista = [10, 20, 30, 40, 50]
mi_tupla = ("a", "b", "c", "d")
mi_cadena = "Python"
```

```python
print(mi_lista[0])   # 10 (primer elemento)
print(mi_tupla[2])    # "c" (tercer elemento)
print(mi_cadena[3])   # "h" (cuarta letra)
```

```python
# 2. Acceso por índice negativo (cuenta desde el final)
print(mi_lista[-1])   # 50 (último elemento)
print(mi_tupla[-2])   # "c" (penúltimo)
print(mi_cadena[-6])  # "P" (primer carácter con índice negativo)
```

```python
# 3. Slicing o rebanado: [inicio:fin]
# Extrae una subparte de la secuencia desde el índice 'inicio' hasta 'fin - 1'
print(mi_lista[1:4])     # [20, 30, 40]
print(mi_tupla[:3])      # ("a", "b", "c")
print(mi_cadena[2:])     # "thon"
```

```python
# 4. Slicing con paso: [inicio:fin:paso]
print(mi_lista[::2])     # [10, 30, 50] (de 2 en 2)
print(mi_cadena[::-1])   # "nohtyP" (cadena invertida)
```

```python
# 5. Indexación anidada (listas dentro de listas)
matriz = [[1, 2], [3, 4], [5, 6]]
print(matriz[0][1])  # 2 (primera fila, segundo valor)
print(matriz[2][0])  # 5 (tercera fila, primer valor)
```

```python
# Métodos útiles para estructuras indexables
# - len(secuencia): longitud total
# - in: para verificar si un valor está presente
# - index(): devuelve el índice de la primera aparición

print(len(mi_lista))        # 5
print("b" in mi_tupla)      # True
print(mi_lista.index(30))   # 2
```

7.Sistemas de Ecuaciones:
Resolver un sistema de ecuaciones lineales
Dado el siguiente sistema de ecuaciones:


3x+2y=12

x−y=1

```python
from numpy.linalg import solve
# Matriz de coeficientes
a=np.array([[3,2],[1,-1]])

# Vector de términos independientes
b=np.array([12,1])
```

```python
# Resolver el sistema
solucion=solve(a,b)
solucion
```

```python
print(f'valor de x:{solucion[0]:.2f}, valor de y:{solucion[1]:.2f}')
```

8.Resolver un sistema de tres ecuaciones
Resuelve el siguiente sistema de ecuaciones:


x+y+z=6

2x+y−z=3

x−y+2z=7

```python
# Matriz de coeficientes
from numpy.linalg import solve
# Matriz de coeficientes
a=np.array([[1,1,1],[2,1,-1],[1,-1,2]])

# Vector de términos independientes
b=np.array([6,3,7])

```

```python
# Resolver el sistema
solucion=solve(a,b)
solucion
print(f'valor de x:{solucion[0]:.2f}, valor de y:{solucion[1]:.2f}, valor de z:{solucion[2]:.2f}')
```

# Ejemplo: ¿Cuál es la altura promedio de los presidentes de Estados Unidos?
Los array disponibles en NumPy pueden resultar extremadamente útiles para resumir un conjunto de valores.
Como ejemplo sencillo, consideremos las alturas de todos los presidentes de Estados Unidos.
Estos datos están disponibles en el archivo *president_heights.csv*, que es una lista simple de etiquetas y valores separados por comas:

```python
# Importamos las librerías necesarias
import pandas as pd
import numpy as np
```

```python
# Creamos una variable donde leemos el archivo en este caso formato csv
df=pd.read_csv('president_heights.csv')
df.head(10)
```

```python
# Tomamos de la tabla la columna de height(cm) y la convertimos en un array
alturas=np.array(df['height(cm)'])
alturas
```

Ahora que tenemos esta matriz de datos, podemos calcular una variedad de estadísticas resumidas:

```python
print(f'Altura Media {alturas.mean():.2f}')
print(f'Altura Maxima {alturas.max():.2f}')
print(f'Altura Minima {alturas.min():.2f}')
print(f'Desviacion estandr en Altura: {alturas.std():.2f}')
```

Otros cálculos que podermos realizar:

```python
Percentil_25 = np.percentile(alturas,25)
Percentil_75 = np.percentile(alturas,75)
Mediana = np.median(alturas)

print(f'Percentil_25 {Percentil_25}')
print(f'Percentil_75 {Percentil_75}')
print(f'Mediana {Mediana}')
```

¿Qué tan selecto es el club de presidentes que superan los 185 cm de altura?

Calcula cuántos presidentes pertenecen a este "club" y cuál es su altura media.

```python
alturas = df["height(cm)"].values
club_altos = alturas[alturas > 185]

print("Miembros del club de los altos:", len(club_altos))
print("Altura media del club:", np.mean(club_altos))
```

Se dice que los primeros presidentes eran más altos.
Calcula la altura media de los primeros 10 presidentes y compárala con los últimos 10.

```python
primeros_10 = alturas[:10]
ultimos_10 = alturas[-10:]

print("Media primeros 10:", np.mean(primeros_10))
print("Media últimos 10:", np.mean(ultimos_10))
```

¿Quién está más cerca de la media?

Usando NumPy, encuentra al presidente cuya altura está más cerca de la altura media de todos.

```python
media_altura = np.mean(alturas)
idx_cercano = np.argmin(np.abs(alturas - media_altura)) # índice de la posición donde se encuentra el valor mínimo
presidente_cercano = df.iloc[idx_cercano]["name"]

print(f"Altura media: {media_altura:.2f} cm")
print(f"Presidente más cercano a la media: {presidente_cercano}")
```

Ahora usando el Dataset de votos debemo unificar la informacion de ambos DF

```python
df_votos=pd.read_csv('Votos_Populares_Simulados.csv')
df_votos.head(5)
```

```python
df_combinado = df.merge(df_votos, on="name")
df_combinado.head(10)
```

¿Quién saca más partido de su altura?

Calcula cuántos millones de votos recibe cada presidente por centímetro de altura.
¿Quién es el más "eficiente"? ¿Y el menos?

```python
df_combinado["votos_por_cm"] = df_combinado["votos_millones"] / df_combinado["height(cm)"]

mas_eficiente = df_combinado.loc[df_combinado["votos_por_cm"].idxmax()]
menos_eficiente = df_combinado.loc[df_combinado["votos_por_cm"].idxmin()]

print("Más eficiente:", mas_eficiente["name"], f"{mas_eficiente['votos_por_cm']:.2f} votos/cm")
print("Menos eficiente:", menos_eficiente["name"], f"{menos_eficiente['votos_por_cm']:.2f} votos/cm")
```

Calcula la correlación entre altura y votos usando NumPy. ¿Existe alguna relación?

```python
altura = df_combinado["height(cm)"].values
votos = df_combinado["votos_millones"].values

correlacion = np.corrcoef(altura, votos)[0, 1]
print(f"Correlación altura-votos: {correlacion:.2f}")
```

¿Qué tan desigual es la distribución de votos?

Calcula el coeficiente de variación (CV) de los votos


```python
std = np.std(votos)
mean = np.mean(votos)
cv = std / mean
print(f"Coeficiente de variación de votos: {cv:.2f}")
```

El top 5 presidencial

¿Quiénes son los 5 presidentes con más votos totales?
¿Cuáles son sus alturas? ¿Hay alguna tendencia?

```python
top5 = df_combinado.sort_values("votos_millones", ascending=False).head(5)
print(top5[["name", "votos_millones", "height(cm)"]])
```

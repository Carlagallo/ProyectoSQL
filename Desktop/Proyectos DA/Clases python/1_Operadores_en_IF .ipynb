### Los Operadores en Python: Aritméticos, Asignación, Comparación y Lógicos 

En programación, los operadores son símbolos especiales que realizan operaciones sobre uno o más valores (conocidos como operandos) y producen un resultado. Son los "verbos" de nuestros programas, permitiéndonos manipular datos, tomar decisiones y construir lógicas complejas.

Python, como lenguaje versátil, ofrece una rica colección de operadores que se pueden clasificar en varias categorías. Comprender cada tipo y cómo interactúan es esencial para escribir código efectivo y eficiente.

Vamos a explorar las principales categorías:

- Operadores Aritméticos

- Operadores de Asignación

- Operadores de Comparación (Relacionales)

- Operadores Lógicos



### 1- Operadores Aritméticos

Los operadores aritméticos se utilizan para realizar cálculos matemáticos básicos. Funcionan con tipos de datos numéricos (enteros, flotantes, etc.).

### Cuadro Resumen de Operadores Aritméticos:

| Operador | Nombre             | Descripción                                       | Ejemplo                 | Resultado     |
| :------- | :----------------- | :------------------------------------------------ | :---------------------- | :------------ |
| `+`      | Suma               | Suma dos operandos.                               | `5 + 3`                 | `8`           |
|          |                    |                                                   | `10.5 + 2.0`            | `12.5`        |
| `-`      | Resta              | Resta el segundo operando del primero.            | `10 - 4`                | `6`           |
|          |                    |                                                   | `7.5 - 2.0`             | `5.5`         |
| `*`      | Multiplicación     | Multiplica dos operandos.                         | `6 * 7`                 | `42`          |
|          |                    |                                                   | `2.5 * 4`               | `10.0`        |
| `/`      | División           | Divide el primer operando por el segundo. El resultado siempre es un `float`. | `10 / 2`                | `5.0`         |
|          |                    |                                                   | `7 / 3`                 | `2.333...`    |
| `//`     | División Entera    | Divide el primer operando por el segundo y devuelve la parte entera del resultado (trunca decimales). | `10 // 3`               | `3`           |
|          |                    |                                                   | `-10 // 3`              | `-4`          |
| `%`      | Módulo / Resto     | Devuelve el resto de la división del primer operando por el segundo. | `10 % 3`                | `1`           |
|          |                    |                                                   | `7 % 2`                 | `1`           |
| `**`     | Exponenciación     | Eleva el primer operando a la potencia del segundo. | `2 ** 3` (2^3)          | `8`           |
|          |                    |                                                   | `9 ** 0.5` (raíz cuadrada) | `3.0`         |

### Puntos Clave de Operadores Aritméticos:

- Orden de Operaciones (Precedencia): 
Python sigue el orden estándar de las operaciones matemáticas 
Puedes usar paréntesis () para forzar un orden de evaluación diferente.

- División ( / ): Siempre devuelve un número de punto flotante (float), incluso si la división es exacta.

- División Entera ( // ): Para números positivos, trunca la parte decimal. Para números negativos, el resultado se redondea hacia abajo (hacia menos infinito), lo que puede ser contraintuitivo a veces.

```python
###💻 EJEMPLOS PRÁCTICOS 
# --- Operadores Aritméticos ---
print("--- Operadores Aritméticos ---")
a = 15
b = 4

print(f"Suma (a + b): {a + b}")
print(f"Resta (a - b): {a - b}")
print(f"Multiplicación (a * b): {a * b}")
print(f"División (a / b): {a / b}") # Siempre flotante
print(f"División Entera (a // b): {a // b}") # Trunca decimales
print(f"Módulo (a % b): {a % b}") # Resto de la división
print(f"Exponenciación (a ** 2): {a ** 2}") # a elevado a la 2
```

### 2- Operadores de Asignación

Los operadores de asignación se utilizan para asignar valores a las variables. 

El operador de asignación más básico es =. 

Los otros operadores de asignación son atajos convenientes para realizar una operación aritmética y una asignación en un solo paso.

### Cuadro Resumen de Operadores de Asignación:

| Operador | Ejemplo    | Equivalente a | Descripción                                                           |
| :------- | :--------- | :------------ | :-------------------------------------------------------------------- |
| `=`      | `x = 5`    | `x = 5`       | Asigna el valor del operando de la derecha al de la izquierda.        |
| `+=`     | `x += 3`   | `x = x + 3`   | Suma y asigna.                                                        |
| `-=`     | `x -= 2`   | `x = x - 2`   | Resta y asigna.                                                       |
| `*=`     | `x *= 4`   | `x = x * 4`   | Multiplica y asigna.                                                  |
| `/=`     | `x /= 2`   | `x = x / 2`   | Divide y asigna. El resultado de la división es siempre un `float`. |
| `//=`    | `x //= 3`  | `x = x // 3`  | División entera y asigna.                                             |
| `%=`     | `x %= 5`   | `x = x % 5`   | Módulo (resto de la división) y asigna.                               |
| `**=`    | `x **= 2`  | `x = x ** 2`  | Exponenciación y asigna.                                              |


### Puntos Clave de Operadores de Asignación:

- Sintaxis Concisa: 

Hacen el código más corto y a menudo más legible, especialmente en operaciones repetitivas sobre una misma variable.

- Variable Existente:

Para los operadores combinados (como +=), la variable de la izquierda (x en los ejemplos) debe ya existir y tener un valor asignado antes de poder usar el operador.

```python
# --- Operadores de Asignación ---
print("\n--- Operadores de Asignación ---")
x = 10
print(f"Valor inicial de x: {x}")

x += 5 # Equivalente a x = x + 5
print(f"x después de x += 5: {x}")

x -= 3 # Equivalente a x = x - 3
print(f"x después de x -= 3: {x}")

x *= 2 # Equivalente a x = x * 2
print(f"x después de x *= 2: {x}")

x /= 4 # Equivalente a x = x / 4
print(f"x después de x /= 4: {x}")

x = 10 # Reiniciamos x para otros ejemplos
x //= 3 # Equivalente a x = x // 3
print(f"x después de x //= 3: {x}")

x = 10
x %= 3 # Equivalente a x = x % 3
print(f"x después de x %= 3: {x}")

x = 2
x **= 3 # Equivalente a x = x ** 3
print(f"x después de x **= 3: {x}")
```

### 3- Operadores de Comparación (Relacionales)

Los operadores de comparación (también conocidos como operadores relacionales) se utilizan para comparar dos valores. El resultado de una comparación siempre es un valor booleano: True (verdadero) o False (falso). Estos operadores son la base de cualquier condición en una sentencia if, elif o else.

### Cuadro Resumen de Operadores de Comparación:

| Operador | Nombre                 | Descripción                                                    | Ejemplo (resulta en `True` o `False`) |
| :------- | :--------------------- | :------------------------------------------------------------- | :------------------------------------ |
| `==`     | Igual a                | `True` si los dos operandos son iguales.                      | `5 == 5` (True)                       |
|          |                        |                                                                | `'hola' == 'adios'` (False)           |
| `!=`     | Diferente de / No igual a | `True` si los dos operandos **NO** son iguales.                 | `10 != 5` (True)                      |
|          |                        |                                                                | `'Python' != 'Python'` (False)        |
| `>`      | Mayor que              | `True` si el operando de la izquierda es mayor que el de la derecha. | `7 > 3` (True)                        |
|          |                        |                                                                | `2 > 8` (False)                       |
| `<`      | Menor que              | `True` si el operando de la izquierda es menor que el de la derecha. | `4 < 9` (True)                        |
|          |                        |                                                                | `15 < 10` (False)                     |
| `>=`     | Mayor o igual que      | `True` si el operando de la izquierda es mayor o igual que el de la derecha. | `6 >= 6` (True)                       |
|          |                        |                                                                | `5 >= 10` (False)                     |
| `<=`     | Menor o igual que      | `True` si el operando de la izquierda es menor o igual que el de la derecha. | `2 <= 2` (True)                       |
|          |                        |                                                                | `7 <= 5` (False)                      |


### Puntos Clave sobre los Operadores de Comparación:

- Tipos de Datos: 

Pueden comparar números, cadenas de texto, y otros tipos de datos. 

Para cadenas, la comparación es lexicográfica (orden alfabético basado en el valor Unicode de los caracteres). 

Listas y tuplas se comparan elemento por elemento.

- Resultados Booleanos: 

Siempre producen True o False.

- is vs == (Opcional pero Importante):

== compara el valor de los objetos.

is compara si dos variables se refieren al mismo objeto en memoria. Son diferentes. Para la mayoría de las comparaciones de valores, == es lo que necesitas.

```python
# --- Operadores de Comparación ---
print("\n--- Operadores de Comparación ---")
num1 = 20
num2 = 25
cadena1 = "Python"
cadena2 = "python" # Diferente por mayúscula

print(f"num1 == num2: {num1 == num2}") # False
print(f"num1 != num2: {num1 != num2}") # True
print(f"num1 > num2: {num1 > num2}")   # False
print(f"num1 < num2: {num1 < num2}")   # True
print(f"num1 >= 20: {num1 >= 20}")     # True
print(f"num2 <= 25: {num2 <= 25}")     # True
print(f"cadena1 == cadena2: {cadena1 == cadena2}") # False (distingue mayúsculas)
print(f"cadena1 != cadena2: {cadena1 != cadena2}") # True
```

### 4. Operadores Lógicos

Los operadores lógicos se utilizan para combinar múltiples expresiones booleanas (True o False) y construir condiciones más complejas. 

Son fundamentales para crear condicionales que evalúen combinaciones de criterios.  

Python tiene tres operadores lógicos principales: and, or, y not.

### Cuadro Resumen de Operadores Lógicos:

| Operador | Nombre       | Descripción                                                                 | Ejemplos (resultan en `True` o `False`)                 |
| :------- | :----------- | :-------------------------------------------------------------------------- | :------------------------------------------------------- |
| `and`    | Y lógico     | Es `True` si **TODAS** las condiciones combinadas son `True`. Si una es `False`, el resultado es `False`. | `(5 > 3) and (10 < 20)`  (True)             |
|          |              |                                                                             | `(2 == 2) and (3 != 3)`  (False)             |
| `or`     | O lógico     | Es `True` si **AL MENOS UNA** de las condiciones combinadas es `True`. Solo es `False` si todas son `False`. | `("a" == "a") or (10 > 100)` (True)          |
|          |              |                                                                             | `(False) or (False)`     (False)             |
| `not`    | Negación lógica | Invierte el valor booleano de la condición. `True` se convierte en `False`, y `False` en `True`. | `not (5 > 10)`           (True)             |
|          |              |                                                                             | `not True`               (False)            |


### Explicación Detallada y Evaluación de Cortocircuito:

- and (Y lógico):

Evaluación de Cortocircuito: Si la primera condición a la izquierda de and se evalúa como False, Python no se molesta en evaluar la segunda condición, porque el resultado de toda la expresión ya se sabe que será False. Esto puede ser útil para optimizar el rendimiento o para evitar errores (ej., if objeto is not None and objeto.metodo(): donde objeto.metodo() no se llamará si objeto es None).
Uso Común: Para asegurar que múltiples criterios se cumplen simultáneamente.

- or (O lógico):

Evaluación de Cortocircuito: Si la primera condición a la izquierda de or se evalúa como True, Python no evalúa la segunda condición, porque el resultado de toda la expresión ya se sabe que será True.
Uso Común: Para permitir que el código se ejecute si se cumple cualquiera de varias condiciones posibles.

- not (Negación lógica):

Uso Común: Para comprobar lo contrario de una condición o para mejorar la legibilidad en ciertos escenarios (ej., if not esta_vacio: es a menudo más claro que if esta_vacio == False:).

```python
# --- Operadores Lógicos ---
print("\n--- Operadores Lógicos ---")
edad = 22
ingresos = 50000
tiene_antecedentes = False
```

```python

# Ejemplo con 'and'
if edad >= 18 and ingresos >= 40000:
    print("Cumple con requisitos de edad e ingresos.")
else:
    print("No cumple con requisitos de edad e ingresos.")
```

```python
# Ejemplo con 'or'
if edad < 18 or tiene_antecedentes:
    print("No apto por edad o antecedentes.")
else:
    print("Apto por edad y sin antecedentes.")
```

```python
# Ejemplo con 'not'
es_miembro_vip = False
if not es_miembro_vip:
    print("No es miembro VIP.")
else:
    print("Es miembro VIP.")
```

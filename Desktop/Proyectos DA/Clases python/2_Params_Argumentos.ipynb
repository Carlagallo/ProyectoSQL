### Parámetros y Argumentos en Funciones (Params & Args):

Ya vimos que es una funcion y como se define, continuemos analizando las estructuras de las funciones

### 📌 Parámetros vs Argumentos

| Concepto   | ¿Qué es?                                                                      |
|------------|--------------------------------------------------------------------------------|
| Parámetro  | Variable que se **define** en la función. Es una "entrada esperada".          |
| Argumento  | Valor **real** que se **pasa** a la función cuando se llama.                  |


📌 Los parámetros están en la definición de la función,
🧾 Los argumentos están en la llamada a la función.


## Tipos de parámetros

- Posicionales: Se pasan en orden.

- Nombrados o keywords: Se indica el nombre del parámetro al llamarlo.

- Por defecto: Tienen un valor si no se especifica.

- Variables (*args, **kwargs): Para recibir muchos valores.

| Tipo                | Sintaxis                         | Descripción                                           |
|---------------------|----------------------------------|-------------------------------------------------------|
| Posicional          | `funcion(5, 10)`                 | Se pasan en orden según la definición                 |
| Por defecto         | `def saludo(nombre="Amigo")`     | Si no se da valor, usa el valor predeterminado        |
| Nombrado (keyword)  | `funcion(b=10, a=5)`             | Se indica cada argumento con su nombre                |
| *args               | `def sumar(*numeros)`            | Recibe varios argumentos como tupla                   |
| **kwargs            | `def mostrar(**info)`            | Recibe argumentos nombrados como diccionario          |


### Veamos algunos ejemplos:

### Ejemplo 1: Parámetros posicionales

```python
def sumar(a, b):
    return a + b

print(sumar(3, 5))
```

### Ejemplo 2: Parámetros con valor por defecto

```python
def saludar(nombre="invitado"):
    print(f"Hola, {nombre}")

saludar()          
saludar("Marta")  
```

### Ejemplo 3: Argumentos nombrados

```python
def presentar(nombre, edad):
    print(f"{nombre} tiene {edad} años")

presentar(edad=30, nombre="Luis")
```

### Ejemplo 4: *args (múltiples valores posicionales)

```python
def sumar_todo(*numeros):
    total = 0
    for n in numeros:
        total += n
    return total

print(sumar_todo(1,2,3,4,6,10))
```

### Ejemplo 5: **kwargs (múltiples valores nombrados)


```python
def mostrar_info(**kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

mostrar_info(nombre="Ana", edad=25, ciudad="Madrid" , Genero = 'Femenino', profesion= 'Diseñador')
```

### Variables Locales y Globales

¿Qué es el scope (alcance)?

El scope define desde dónde se puede acceder a una variable. En Python, hay principalmente dos tipos:

| Tipo   | ¿Dónde se define?             | ¿Desde dónde se accede?                             |
|--------|-------------------------------|-----------------------------------------------------|
|🔒Local  | Dentro de una función         | Solo se usa dentro de esa función                   |
| 🌍Global | Fuera de todas las funciones  | Se puede usar desde cualquier parte del código      |

### Veamos algunos ejemplos:

### Ejemplo: Variable local:



```python
def saludo():
    mensaje = "Hola"  # variable local
    print(mensaje)

saludo()  
```

### Ejemplo: Variable global:

```python
mensaje = "Hola"  # variable global

def saludo():
    print(mensaje)

saludo()        
print(mensaje) 
```

¿Y si quiero modificar una global dentro de una función?

Necesitas usar la palabra clave global, con cuidado:

```python
contador = 0
print(f'Al inicio {contador}')

def aumentar():
    global contador
    contador += 5

aumentar()
print(f'Modificado {contador}')
```

```python
contador = 0
print(f'Al inicio {contador}')

def aumentar():
    #global contador
    contador += 5

aumentar()
print(f'Modificado {contador}')
```

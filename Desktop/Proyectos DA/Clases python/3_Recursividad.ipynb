### 🌀 Recursividad en Python:

### ¿Qué es la recursividad?

La recursividad es una técnica de programación donde una función se llama a sí misma para resolver un problema.

Se basa en dividir un problema grande en subproblemas más pequeños, hasta llegar a un caso base que detiene la recursión.

### Componentes clave:

- Caso base: 
    condición que detiene la recursión.

- Llamada recursiva: 
    la función se llama a sí misma con un problema más pequeño.

### 🔹 ¿Cuándo usarla?

Cuando un problema se puede dividir naturalmente en partes más pequeñas similares al original.

Algunos ejemplos típicos: factorial, secuencia de Fibonacci, recorrido de estructuras anidadas, etc.

### 🔥 ¡Cuidado!

Cada llamada recursiva ocupa espacio en la pila de llamadas (stack).

Un mal diseño puede llevar a un RecursionError.


### 🔁 Recursividad: Resumen clave

| Elemento           | Descripción                                               |
|--------------------|-----------------------------------------------------------|
| Función recursiva  | Se llama a sí misma                                       |
| Caso base          | Condición para detener la recursión                       |
| Llamada recursiva  | Reducción del problema hacia el caso base                 |
| Riesgos            | Stack overflow si no se detiene, alto consumo de memoria  |
| Alternativa        | Usar bucles `for` o `while` si se puede                   |

### Vamos a ver algunos Ejemplos
---

### Ejemplo 1: Factorial de un número



````python
def factorial(n):
    if n == 0 or n == 1:  # 🧱 Caso base
        return 1
    else:
        return n * factorial(n - 1)  # 🔁 Llamada recursiva

print(factorial(10))
````

### Ejemplo 2: Suma de los primeros n números

````python
def suma_recursiva(n):
    if n == 0:
        return 0
    else:
        return n + suma_recursiva(n - 1)

print(suma_recursiva(4))
````

### Ejemplo 3: Contar regresivamente

````python
def cuenta_regresiva(n):
    if n == 0:
        print("¡Despegue!")
    else:
        print(n)
        cuenta_regresiva(n - 1)

cuenta_regresiva(5)
````

### Ejemplo 4: Fibonacci (versión básica)

````python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

print(fibonacci(6)) 
````

### Ejercicio 5
Crea una función recursiva que determine si una palabra es un palíndromo (se lee igual al derecho y al revés).
Ejemplo: `es_palindromo("radar")` ➜ True

### como funciona esto???

La función `es_palindromo(palabra)` verifica si una palabra se lee igual al derecho y al revés.

### 🔹 Caso base:

🔽 Si la palabra tiene **0 o 1 letras**, se considera un palíndromo.  
  
```python
if len(palabra) <= 1:
    return True
```

### 🔹 Comparación de extremos:

Si la primera letra (palabra[0]) es diferente de la última (palabra[-1])

→ ❌ No es palíndromo → return False


### 🔹 Llamada recursiva:
Si los extremos coinciden, se repite el proceso con el interior de la palabra:

return es_palindromo(palabra[1:-1])


🔁 PASO A PASO

🟢 Ejemplo: es_palindromo("radar")





| Llamada                   | palabra | palabra\[0] | palabra\[-1] | Comparación | Resultado                          |
|---------------------------|---------|-------------|--------------|-------------|-------------------------------------|
| `es_palindromo("radar")`  | radar   | r           | r            | ✅ iguales  | sigue con `es_palindromo("ada")`   |
| `es_palindromo("ada")`    | ada     | a           | a            | ✅ iguales  | sigue con `es_palindromo("d")`     |
| `es_palindromo("d")`      | d       | -           | -            | caso base   | ✅ return True                      |

🔚 Resultado final: True

🔴 Ejemplo: es_palindromo("python")

| Llamada                   | palabra | palabra\[0] | palabra\[-1] | Comparación | Resultado    |
| ------------------------- | ------- | ----------- | ------------ | ----------- | ------------ |
| `es_palindromo("python")` | python  | p           | n            | ❌ distintas | return False |


🔚 Resultado final: False

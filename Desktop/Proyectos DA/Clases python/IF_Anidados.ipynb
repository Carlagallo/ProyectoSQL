### Condicionales if Anidados 

Los condicionales anidados son una herramienta poderosa en la programación que nos permite manejar decisiones complejas que dependen de múltiples factores evaluados en un orden jerárquico. Un if anidado es, simplemente, una sentencia if (o if-else, if-elif-else) que se encuentra dentro del bloque de código de otra sentencia condicional (ya sea un if, elif o else).

Imagina que estás en un proceso de toma de decisiones donde la siguiente pregunta que haces depende de la respuesta a la pregunta anterior. Por ejemplo:

"Si el tiempo es soleado, entonces (y solo entonces) revisa si tienes planes de playa. Si sí, ve a la playa; si no, ve al parque."

"Si el usuario está autenticado, entonces (y solo entonces) verifica su nivel de permisos. Si es administrador, accede a configuración; si es usuario normal, accede a su perfil."
Concepto Clave: Jerarquía y Flujo de Ejecución

La característica definitoria de los if anidados es su estructura jerárquica. La condición del if interno solo será evaluada por Python si se cumple la condición del if (o elif/else) externo que lo contiene. Si la condición externa es False y no se llega al bloque donde está anidado, el if interno nunca se ejecutará.

Sintaxis Básica de Anidación:

```Python
if condicion_externa:
    # Este bloque se ejecuta si 'condicion_externa' es True

    if condicion_interna_1:
        # Este código se ejecuta SOLO si 'condicion_externa' es True
        # Y 'condicion_interna_1' es True
        print("Ambas condiciones son verdaderas")
    else:
        # Este código se ejecuta SOLO si 'condicion_externa' es True
        # PERO 'condicion_interna_1' es False
        print("La condición externa es verdadera, pero la interna es falsa")

else:
    # Este bloque se ejecuta si 'condicion_externa' es False
    print("La condición externa es falsa, las internas no se evalúan aquí.")
    # Puedes tener más anidación aquí si es necesario
    # if otra_condicion:
    #     pass
```

### La Indentación: ¡Tu Mejor Amiga (o Peor Enemiga)!

En Python, la indentación no es solo para la legibilidad; 
es parte de la sintaxis del lenguaje y define qué líneas de código pertenecen a qué bloque. 

Para los if anidados, esto es aún más crítico:

Cada nivel de anidación requiere un nivel adicional de indentación.

Por convención, se usan cuatro espacios para cada nivel de indentación.

Si una línea no está correctamente indentada, Python interpretará que no forma parte del bloque anidado o, en el peor de los casos, generará un IndentationError.

### Ventajas y Desventajas de la Anidación:

- Ventajas:
Permite modelar lógicas de decisión complejas y ramificadas de forma clara cuando las decisiones son secuenciales.
Ideal para escenarios donde una validación es un pre-requisito para la siguiente (ej., verificar usuario antes de contraseña).

- Desventajas:
"Pirámide de la Muerte" (Arrowhead Anti-pattern): Demasiados niveles de anidación pueden hacer que el código sea muy difícil de leer, entender y mantener. Visualmente, el código se desplaza mucho hacia la derecha.

A menudo, se pueden reemplazar anidaciones profundas con un uso inteligente de operadores lógicos (and, or, not), lo que resulta en un código más plano y legible.
Es fundamental dominar ambos enfoques para escribir código Python eficiente y legible.

````python
# --- Ejemplo 1: Verificación de Elegibilidad para un Préstamo ---
print("--- Ejemplo 1: Verificación de Elegibilidad para un Préstamo ---")
ingreso_mensual = 3500  # Euros
historial_credito_bueno = True
antiguedad_empleo_años = 2

if ingreso_mensual >= 3000 :
    print("Cumple con el requisito de ingresos.")
    if historial_credito_bueno:
        print("Cumple con el requisito de historial crediticio.")
        if antiguedad_empleo_años >= 3:
            print("¡Felicidades! Es elegible para el préstamo.")
        else:
            print("No es elegible: Le falta antigüedad en el empleo.")
    else:
        print("No es elegible: Historial crediticio no es bueno.")
else:
    print("No es elegible: Ingresos mensuales insuficientes.")


````

````python
# --- Ejemplo 2: Categorización de Edad y Nivel de Educación ---
print("\n--- Ejemplo 2: Categorización de Edad y Nivel de Educación ---")
edad = 25
nivel_educacion = "universitario" # "primaria", "secundaria", "universitario", "postgrado"

if edad < 18:
    print("Es menor de edad.")
    if nivel_educacion == "primaria":
        print("Está en etapa de educación primaria o inicial.")
    else:
        print("Nivel de educación inesperado para su edad (¿quizás pre-universitario precoz?).")
elif edad >= 18 and edad <= 65: # Rango de edad adulta
    print("Es un adulto.")
    if nivel_educacion == "universitario" or nivel_educacion == "postgrado":
        print("Profesional con educación superior.")
    elif nivel_educacion == "secundaria":
        print("Adulto con educación secundaria.")
    else:
        print("Adulto con otro nivel de educación.")
else: # edad > 65
    print("Es un adulto mayor.")
    if nivel_educacion == "postgrado":
        print("Adulto mayor con alta formación académica.")
    else:
        print("Adulto mayor.")


````

````python
# --- Ejemplo 3: Menú Interactivo Simple (simulando opciones) ---
print("\n--- Ejemplo 3: Menú Interactivo Simple ---")
opcion_principal = input("Elija una opción (1: Comprar, 2: Vender, 3: Salir): ")

if opcion_principal == '1':
    print("Ha seleccionado Comprar.")
    tipo_producto = input("¿Qué tipo de producto desea? (A: Electrónica, B: Ropa): ")
    if tipo_producto.upper() == 'A':
        print("Mostrando productos de Electrónica.")
    elif tipo_producto.upper() == 'B':
        print("Mostrando productos de Ropa.")
    else:
        print("Tipo de producto no válido.")
elif opcion_principal == '2':
    print("Ha seleccionado Vender.")
    confirmacion = input("¿Está seguro de querer vender? (si/no): ")
    if confirmacion.lower() == 'si':
        print("Accediendo a la sección de venta de artículos.")
    else:
        print("Operación de venta cancelada.")
elif opcion_principal == '3':
    print("Saliendo del programa. ¡Hasta pronto!")
else:
    print("Opción principal no válida.")
````
